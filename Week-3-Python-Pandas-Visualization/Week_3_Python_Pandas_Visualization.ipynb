```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df.index, df["Calories"])

plt.title("Calories Burned Across Workouts")
plt.xlabel("Workout Record")
plt.ylabel("Calories")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Duration"], df["Calories"])

plt.title("Workout Duration vs Calories Burned")
plt.xlabel("Duration")
plt.ylabel("Calories")

plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Calories"],
    kde=True
)

plt.title("Distribution of Calories Burned")
plt.xlabel("Calories")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["Calories"]
)

plt.title("Distribution of Calories Burned")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = df[
    ["Duration", "Pulse", "Maxpulse", "Calories"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Workout Variables")

plt.show()
```

```python
# Creating a lookup table for workout duration

duration_lookup = pd.DataFrame({
    "Duration": sorted(df["Duration"].dropna().unique())
})

duration_lookup["Session_Type"] = pd.cut(
    duration_lookup["Duration"],
    bins=[-float("inf"), 45, 60, float("inf")],
    labels=["Short", "Medium", "Long"]
)

duration_lookup
```

```python
# Merging the original dataset with the duration lookup table

merged_df = pd.merge(
    df,
    duration_lookup,
    on="Duration",
    how="left"
)

merged_df.head()
```

```python
merged_df[["Duration", "Calories", "Session_Type"]].head(10)
```

```python
merged_df.groupby("Session_Type")["Calories"].mean().round(2)
```
